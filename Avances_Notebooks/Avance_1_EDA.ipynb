### Y que pasa con las inconsistencias en cantidad de visitas

```python
# Revisemos que tipo de visitas quedan por corregir
# Definir condiciones a revisar
condiciones = {
    "Visitas = 0": df_chicago_clean['frecuencia_visita'] == 0,
    "Visitas negativas": df_chicago_clean['frecuencia_visita'] < 0
}

# Recorrer y mostrar resultados
for nombre, condicion in condiciones.items():
    registros = df_chicago_clean[condicion]
    print(f"{nombre}: {len(registros)} registros")
    print("Distribución de frecuencia_visita:")
    print(registros['frecuencia_visita'].value_counts().sort_index())
    print("-" * 40)
```

### En el dataset se identificaron registros con valores problemáticos en la columna frecuencia_visita:
Valores negativos (-3): es probable que sea un error de registro. Si los registros tienen datos completos, como promedio_gasto_comida distinto de $0, id, nombre, apellido, género y al menos teléfono o correo, se consideran clientes válidos.

Se decidió:
No borrar registros.
Corregir los negativos tomando el valor absoluto (-3 → 3).

```python
#Procedemos a corregir los valores negativos (Error de tipeo: -3 -> 3)

# Identificar registros con frecuencia negativa
negativos = df_chicago_clean[df_chicago_clean['frecuencia_visita'] < 0]
print(f"Registros con frecuencia negativa (error): {len(negativos)}")

if len(negativos) > 0:
    # Aplicar valor absoluto para corregir el error
    df_chicago_clean.loc[negativos.index, 'frecuencia_visita'] = negativos['frecuencia_visita'].abs()
    print(f"Valores negativos corregidos: {len(negativos)} registros (ej: -3 → 3)")
else:
    print("No se encontraron valores negativos.")

# Verificamos
print(f"\nDistribucion final de 'frecuencia_visita':")
distribucion_final = df_chicago_clean['frecuencia_visita'].value_counts().sort_index()
print(distribucion_final)
```

```python
# Para entender mejor que imputar, vamos con un analisis cruzado de datos
#Revisemos por genero
print("Distribucion de preferencias por genero:")
preferencias_por_genero = pd.crosstab(df_chicago_clean['genero'], 
                                     df_chicago_clean['preferencias_alimenticias'])
print(preferencias_por_genero)
```

```python
# Ahora analicemos por grupos de edad
# Crear grupos de edad
df_chicago_clean['grupo_edad'] = pd.cut(df_chicago_clean['edad'], 
                                         bins=[0, 25, 35, 45, 55, 100], 
                                         labels=['18-25', '26-35', '36-45', '46-55', '56+'])

print("\nDistribucion de preferencias por grupo de edad:")
preferencias_por_edad = pd.crosstab(df_chicago_clean['grupo_edad'], 
                                   df_chicago_clean['preferencias_alimenticias'])
print(preferencias_por_edad)
```

```python
# Estrategia de imputacien inteligente basada en genero y grupo de edad

# Primero creemos grupos de edad mes detallados
df_chicago_clean['grupo_edad'] = pd.cut(df_chicago_clean['edad'], 
                                         bins=[0, 25, 35, 45, 55, 65, 100], 
                                         labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+'])

# Funcien para imputar basado en genero y grupo de edad
def imputar_preferencia_inteligente(fila):
    if pd.isna(fila['preferencias_alimenticias']):
        try:
            # Buscar personas del mismo genero y grupo de edad
            mismo_grupo = df_chicago_clean[
                (df_chicago_clean['genero'] == fila['genero']) & 
                (df_chicago_clean['grupo_edad'] == fila['grupo_edad'])
            ]
            
            if len(mismo_grupo) > 0:
                # Calcular la moda del subgrupo
                moda_grupo = mismo_grupo['preferencias_alimenticias'].mode()
                if len(moda_grupo) > 0:
                    return moda_grupo[0]
            
            # Si no hay suficientes datos en el subgrupo, usar solo genero
            mismo_genero = df_chicago_clean[df_chicago_clean['genero'] == fila['genero']]
            moda_genero = mismo_genero['preferencias_alimenticias'].mode()
            if len(moda_genero) > 0:
                return moda_genero[0]
                
        except:
            # En caso de error, usar la moda general
            pass
            
    return fila['preferencias_alimenticias']

# Aplicar la imputacion
print("Aplicando imputacion inteligente...")
nulos_antes = df_chicago_clean['preferencias_alimenticias'].isnull().sum()

df_chicago_clean['preferencias_alimenticias'] = df_chicago_clean.apply(
    imputar_preferencia_inteligente, axis=1
)

nulos_despues = df_chicago_clean['preferencias_alimenticias'].isnull().sum()
print(f"Valores nulos imputados: {nulos_antes - nulos_despues}")


# Verificacion final
nulos_final = df_chicago_clean['preferencias_alimenticias'].isnull().sum()
print(f"Valores nulos finales: {nulos_final}")

# Eliminar columna temporal
df_chicago_clean = df_chicago_clean.drop('grupo_edad', axis=1)

# Mostrar distribucion final
print("\nDistribucion final de preferencias alimenticias:")
print(df_chicago_clean['preferencias_alimenticias'].value_counts())
```

### OPORTUNIDADES DE COMPLETADO 
#### Hay 192 registros duplicados (360 registros en total)
#### De estos, 191 tienen teléfono nulo y 181 tienen correo nulo

```python
# Funcion optimizada para completar datos
def completar_datos_grupo(grupo):
    for columna in ['telefono_contacto', 'correo_electronico']:
        valores_no_nulos = grupo[columna].dropna()
        if len(valores_no_nulos) > 0:
            grupo[columna] = grupo[columna].fillna(valores_no_nulos.iloc[0])
    return grupo

# Aplicar completado de datos
df_chicago_completado = df_chicago_clean.groupby('nombre_completo', group_keys=False).apply(completar_datos_grupo)

# Resultados concisos
nulos_despues = df_chicago_completado[['telefono_contacto', 'correo_electronico']].isnull().sum()
completados_telefono = 2584 - nulos_despues['telefono_contacto']
completados_correo = 2547 - nulos_despues['correo_electronico']

print(f"Datos completados: Telefonos: {completados_telefono} | Correos: {completados_correo}")
```

```python
# Completar valores nulos restantes con "no completo"
df_chicago_completado['telefono_contacto'] = df_chicago_completado['telefono_contacto'].fillna('no completo')
df_chicago_completado['correo_electronico'] = df_chicago_completado['correo_electronico'].fillna('no completo')

# Guardar el dataset limpio y procesado en la carpeta Archivos CSV
# ../Archivos_CSV/ -> sube un nivel (del notebook) y entra en la carpeta Archivos CSV
ruta_guardado = "../Archivos_CSV/clientes_chicago_procesado.csv"
df_chicago_completado.to_csv(ruta_guardado, index=False)
```
